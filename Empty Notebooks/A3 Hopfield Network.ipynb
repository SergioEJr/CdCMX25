# MTY4-A3: Hopfield Network

A Hopfield network is a recurrent neural network whose dynamics and energy function are directly analogous to the Ising model. In the context of Hopfield networks, the spins are called **neurons** and they are responsible for storing and retrieve patterns. The model is a simple example of associative memory (also known as content-aware memory), where stored patterns are attractor states of the dynamics. In other words, the particular patterns that are stored in the model *minimize the energy*, so the system naturally flows towards these states.

The *Hamiltonian* of the Hopfield model looks very similar to the Ising model,

$$H[\{\sigma_i\}] = - \sum_{i j} K_{ij} \sigma_i \sigma_j$$

The two main differences in the Hopfield network are:
- The temperature is set to 0 -- there are no thermal fluctuations so the only thing driving the system is **energy minimization**
- The lattice is **fully connected** -- each neuron is connected to every other neuron

We update the state as follows:

1. Randomly select a neuron $\sigma_i$
2. Calculate the change in the energy $\Delta H$ from flipping the neuron $\sigma_i \to -\sigma_i$
3. Determine whether to accept the spin flip or not\
   a. If the energy is decreased, $\Delta H < 0$, accept the spin flip\
   b. If the energy is increased, $\Delta H > 0$, do nothing
