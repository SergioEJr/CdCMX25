The Ising model is a foundational model in the field of statistical physics -- it is the *drosophilia* of statistical mechanics. It is the canonical example in the study of **phase transitions** and serves as an elementary model for magnetism. The model consists of a 2D array of *spins* $\sigma_i = \pm 1$. These are tiny little magnets that either point up or down. Nearby spins like to be oriented in the same direction. This behavior is encompassed in the system's *Hamiltonian* -- otherwise known as the **energy**. A given spin configuration (microstate) $\{\sigma_i \}$ has energy,

$$H[\{\sigma_i\}] = - \sum_{\langle i j \rangle} K_{ij} \sigma_i \sigma_j$$

The numbers $K_{ij}$ quantify how strong the interaction is between $\sigma_i$ and $\sigma_j$ and the notation $\langle i j \rangle$ indicates that the sum is over nearest neighbors. For simplicity, we let all interactions be of the same strength $K_{ij} = K$. Notice that if all spins point in the same direction, for example $\sigma_i = 1$ for all $i$, then the sum is maximized and the energy $H$ is minimized. 

We imagine that this lattice is an open system, so that it can exchange energy with the environment. This exchange of energy is quantified by the **temperature** $T$. The thermal fluctuations due to the environment can spontaneously flip spins at random. It is the competition between these thermal fluctuations and the system's desire to minimize it's internal energy that lead to the phase transition.

Notice that this Hamiltonian is symmetric under the transformation $\sigma_i \to - \sigma_i$. In other words, flipping all of the spins leaves us with the same energy. We can break this symmetry by including an *external magnetic field* $h$. Note that spins pointing in the direction of the external field are favored -- they lower the energy --  whereas spins pointing against the field $h$ increase the energy.

$$H[\{\sigma_i\}] = -K \sum_{\langle i j \rangle} \sigma_i \sigma_j - h \sum_i \sigma_i$$

## Simulation

The simulation of the Ising model is fundamentally different than our simulations up to now. All of our previous simulations have been *non-interacting*, each particle could evolve independently of the others. However, in the Ising model neighboring spins **directly influence each other** -- they are not independent. The model is thus much richer and interesting. 

In this lattice simulation we will have a 2D array of size $L\times L$. Each site represents a spin that can either point down (-1) or up (+1). The lattice should be initialized with all spins pointing in random directions.

We update the state of the system as follows:

1. Randomly select a spin $\sigma_i$
2. Calculate the change in the energy $\Delta H$ from flipping the spin $\sigma_i \to -\sigma_i$
3. Determine whether to accept the spin flip or not\
    a. If the energy is decreased, $\Delta H < 0$, accept the spin flip with probability 1 (always accept)\
    b. If the energy is increased, $\Delta H > 0$, accept the spin flip with probability $\exp(-\Delta H/T)$

We will have **periodic boundary conditions** in both the horizontal and vertical directions.

```python
# common parameters between all three simulations
n_sweeps = 1000
L = 75
K = 1
h = 0

# three different temperatures
lowT = 1.5
critT = 2.27
highT = 3.0

# three simulations
lowT_snapshots = sols.snapshot_series(n_sweeps, L, lowT, K, h, prog=True)
critT_snapshots = sols.snapshot_series(n_sweeps, L, critT, K, h, prog=True)
highT_snapshots = sols.snapshot_series(n_sweeps, L, highT, K, h, prog=True)
```

To create multiple axes on the same figure, we specify the number of `(rows, cols)` inside of the `subplots` function. We can then add data to these axes as usual.

```python
fig, axs = plt.subplots(1,3, dpi=250)
titles = ["Low T", "Crit T", "High T"]
lowT = axs[0].imshow(lowT_snapshots[0], cmap="grey")
critT = axs[1].imshow(critT_snapshots[0], cmap="grey")
highT = axs[2].imshow(highT_snapshots[0], cmap="grey")
for i, ax in enumerate(axs):
    ax.set_title(titles[i])
    ax.set_xticks([])
    ax.set_yticks([])
    
# update function for the animation
def update_frame(n):
    lowT.set_array(lowT_snapshots[n])
    critT.set_array(critT_snapshots[n])
    highT.set_array(highT_snapshots[n])
    return lowT, critT, highT
```

```python
anim = animation.FuncAnimation(fig, func=update_frame, frames=len(lowT_snapshots), interval=50, blit=True)
HTML(anim.to_jshtml())
```
